## 01. Linear Search

**Aim:** 1. Write a program in python to search an element using linear search and find its execution time using the time function.

```
Algorithm: LinearSearch(A, n, x)

1. Start
2. For i <- 0 to (n-1) do
3.        if x == A[i] then
4.              print "Element Found"
5.              exit
6. If (i==n) then
7.         print "Element Not Found"
8. Stop
```

````python
import time
n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))
x=int(input("Enter the element to search:"))

st=time.perf_counter()
c,pos=0,-1
for i in range(n):
    c+=1
    if a[i]==x: pos=i; break
t=time.perf_counter()-st

print(f"\n{x} is not found" if pos==-1 else f"\n{x} is found at position {pos+1}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: best case O(1), worst case O(n)\nSpace complexity: O(1)")
````

## 02. Binary Search

**Aim:** 2. Write a program in python to search an element using binary search and find its execution time using the time function.

```
Algorithm: BinarySearch(A, n, x)

1. Start
2. low <- 0, high <- (n-1)
3. While (low <= high) do
4.        mid <- (low+high)/2
5.        if x == A[mid] then
6.              print "Element Found"
7.              exit
8.        else if A[mid] < x then
9.              low <- mid+1
10.       else
11.             high <- mid-1
12. If (low > high) then
13.        print "Element Not Found"
14. Stop
```

````python
import time
n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements in ascending order:").split()))
x=int(input("Enter the element to search:"))

st=time.perf_counter()
l,h,m,c=0,n-1,-1,0
while l<=h:
    c+=1; m=(l+h)//2
    if a[m]==x: break
    elif a[m]<x: l=m+1
    else: h=m-1
t=time.perf_counter()-st

print(f"\n{x} is not found" if l>h else f"\n{x} is found at position {m+1}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: best case O(1), worst case O(log n)\nSpace complexity: O(1)")
````

## 03. Bubble Sort

**Aim:** 3. Write a program in python to sort the elements using bubble sort and find its execution time using the time function.

```
Algorithm: BubbleSort(A, n)

1. Start
2. For i <- 0 to (n-2) do
3.        For j <- 0 to (n-i-2) do
4.              if A[j] > A[j+1] then
5.                    temp <- A[j]
6.                    A[j] <- A[j+1]
7.                    A[j+1] <- temp
8. Print the sorted array A
9. Stop
```

````python
import time
n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))

st=time.perf_counter()
c=0
for i in range(n-1):
    for j in range(n-i-1):
        c+=1
        if a[j]>a[j+1]: a[j],a[j+1]=a[j+1],a[j]
t=time.perf_counter()-st

print("\nAfter bubble sorting:"+"".join(f"\t{v}" for v in a))
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(n^2)\nSpace complexity: O(1)")
````

## 05. Insertion Sort

**Aim:** 5. Write a program in python to sort the elements using insertion sort and find its execution time using the time function.

```
Algorithm: InsertionSort(A, n)

1. Start
2. For i <- 1 to (n-1) do
3.        key <- A[i]
4.        j <- i-1
5.        While (j >= 0 and A[j] > key) do
6.              A[j+1] <- A[j]
7.              j <- j-1
8.        A[j+1] <- key
9. Print the sorted array A
10. Stop
```

````python
import time
n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))

st=time.perf_counter()
c=0
for i in range(1,n):
    key,j=a[i],i-1
    while j>=0:
        c+=1
        if a[j]<=key: break
        a[j+1]=a[j]; j-=1
    a[j+1]=key
t=time.perf_counter()-st

print("\nAfter insertion sorting:"+"".join(f"\t{v}" for v in a))
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: best case O(n), worst case O(n^2)\nSpace complexity: O(1)")
````

## 06. Radix Sort

**Aim:** 6. Write a program in python to sort the elements using radix sort and find its execution time using the time function.

```
Algorithm: RadixSort(A, n)

1. Start
2. max <- largest element of A
3. exp <- 1
4. While (max/exp > 0) do
5.        CountSort(A, n, exp)
6.        exp <- exp*10
7. Print the sorted array A
8. Stop

Algorithm: CountSort(A, n, exp)

1. For i <- 0 to 9 do
2.        bucket[i] <- 0
3. For i <- 0 to (n-1) do
4.        d <- (A[i]/exp) mod 10
5.        bucket[d] <- bucket[d]+1
6. For i <- 1 to 9 do
7.        bucket[i] <- bucket[i]+bucket[i-1]
8. For i <- (n-1) down to 0 do
9.        d <- (A[i]/exp) mod 10
10.       bucket[d] <- bucket[d]-1
11.       output[bucket[d]] <- A[i]
12. For i <- 0 to (n-1) do
13.       A[i] <- output[i]
14. Return
```

````python
import time
c=0

def countSort(a,e):
    global c
    b,o=[0]*10,[0]*len(a)
    for v in a:
        c+=1; b[v//e%10]+=1
    for i in range(1,10): b[i]+=b[i-1]
    for v in reversed(a):
        c+=1; d=v//e%10; b[d]-=1; o[b[d]]=v
    a[:]=o

n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))

st=time.perf_counter()
mx,e=max(a),1
while mx//e>0:
    countSort(a,e); e*=10
t=time.perf_counter()-st

print("\nAfter radix sorting:"+"".join(f"\t{v}" for v in a))
print(f"Number of operations:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(d*n)\nSpace complexity: O(n+k)")
````

## 07. Bucket Sort

**Aim:** 7. Write a program in python to sort the elements using bucket sort and find its execution time using the time function.

```
Algorithm: BucketSort(A, n)

1. Start
2. max <- largest element of A
3. For i <- 0 to (n-1) do
4.        size[i] <- 0
5. For i <- 0 to (n-1) do
6.        index <- (A[i]*n)/(max+1)
7.        bucket[index][size[index]] <- A[i]
8.        size[index] <- size[index]+1
9. For i <- 0 to (n-1) do
10.       sort bucket[i] using insertion sort
11. k <- 0
12. For i <- 0 to (n-1) do
13.       For j <- 0 to (size[i]-1) do
14.             A[k] <- bucket[i][j]
15.             k <- k+1
16. Print the sorted array A

```

````python
import time
n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))

st=time.perf_counter()
mx,c=max(a),0
b=[[] for _ in range(n)]
for x in a: b[x*n//(mx+1)].append(x)
for v in b:
    for i in range(1,len(v)):
        key,j=v[i],i-1
        while j>=0:
            c+=1
            if v[j]<=key: break
            v[j+1]=v[j]; j-=1
        v[j+1]=key
a=[x for v in b for x in v]
t=time.perf_counter()-st

print("\nAfter bucket sorting:"+"".join(f"\t{v}" for v in a))
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: average O(n+k), worst O(n^2)\nSpace complexity: O(n^2)")
````

## 08. Quick Sort

**Aim:** 8. Write a program in python to sort the elements using quick sort and find its execution time using the time function.

```
Algorithm: QuickSort(A, low, high)

1. Start
2. If (low < high) then
3.        p <- Partition(A, low, high)
4.        QuickSort(A, low, p-1)
5.        QuickSort(A, p+1, high)
6. Stop

Algorithm: Partition(A, low, high)

1. pivot <- A[high]
2. i <- low-1
3. For j <- low to (high-1) do
4.        if A[j] <= pivot then
5.              i <- i+1
6.              swap A[i] and A[j]
7. swap A[i+1] and A[high]
8. Return i+1
```

````python
import time
c=0

def partition(a,l,h):
    global c
    i=l-1
    for j in range(l,h):
        c+=1
        if a[j]<=a[h]: i+=1; a[i],a[j]=a[j],a[i]
    a[i+1],a[h]=a[h],a[i+1]
    return i+1

def quickSort(a,l,h):
    if l<h:
        p=partition(a,l,h)
        quickSort(a,l,p-1)
        quickSort(a,p+1,h)

n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))

st=time.perf_counter()
quickSort(a,0,n-1)
t=time.perf_counter()-st

print("\nAfter quick sorting:"+"".join(f"\t{v}" for v in a))
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(n log n) average, O(n^2) worst\nSpace complexity: O(log n) average, O(n) worst")
````

## 09. Merge Sort

**Aim:** 9. Write a program in python to sort the elements using merge sort and find its execution time using the time function.

```
Algorithm: MergeSort(A, low, high)

1. Start
2. If (low < high) then
3.        mid <- (low+high)/2
4.        MergeSort(A, low, mid)
5.        MergeSort(A, mid+1, high)
6.        Merge(A, low, mid, high)
7. Stop

Algorithm: Merge(A, low, mid, high)

1. n1 <- mid-low+1, n2 <- high-mid
2. Copy A[low..mid] into L[0..n1-1] and A[mid+1..high] into R[0..n2-1]
3. i <- 0, j <- 0, k <- low
4. While (i < n1 and j < n2) do
5.        if L[i] <= R[j] then
6.              A[k] <- L[i], i <- i+1
7.        else
8.              A[k] <- R[j], j <- j+1
9.        k <- k+1
10. Copy the remaining elements of L and R into A
11. Return
```

````python
import time
c=0

def merge(L,R):
    global c
    out=[]
    while L and R:
        c+=1
        out.append(L.pop(0) if L[0]<=R[0] else R.pop(0))
    return out+L+R

def mergeSort(a):
    if len(a)<2: return a
    m=(len(a)+1)//2
    return merge(mergeSort(a[:m]),mergeSort(a[m:]))

n=int(input("Enter the number of elements:"))
a=list(map(int,input("Enter the elements:").split()))

st=time.perf_counter()
a=mergeSort(a)
t=time.perf_counter()-st

print("\nAfter merge sorting:"+"".join(f"\t{v}" for v in a))
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(n log n)\nSpace complexity: O(n)")
````

## 10. Fibonacci (Recursion)

**Aim:** 10. Write a program in python to find the fibonacci series using recursion and find its execution time using the time function.

```
Algorithm: Fib(n)

1. Start
2. If (n <= 1) then
3.        Return n
4. Return Fib(n-1) + Fib(n-2)
5. Stop

Algorithm: FibonacciSeries(n)

1. Start
2. For i <- 0 to (n-1) do
3.        print Fib(i)
4. Stop
```

````python
import time
calls=0

def fib(n):
    global calls
    calls+=1
    return n if n<=1 else fib(n-1)+fib(n-2)

n=int(input("Enter the number of terms:"))

st=time.perf_counter()
print("\nFibonacci series:"+"".join(f"\t{fib(i)}" for i in range(n)))
t=time.perf_counter()-st

print(f"Number of function calls:{calls}\nTime taken:{t:f} seconds")
print("Time complexity: O(2^n)\nSpace complexity: O(n)")
````

## 12. Matrix Chain Multiplication

**Aim:** 12. Write a program in python to find the minimum number of scalar multiplications needed for matrix chain multiplication using dynamic programming and find its execution time using the time function.

```
Algorithm: MatrixChainOrder(p, n)

1. Start
2. For i <- 1 to n do
3.        m[i][i] <- 0
4. For l <- 2 to n do
5.        For i <- 1 to (n-l+1) do
6.              j <- i+l-1
7.              m[i][j] <- infinity
8.              For k <- i to (j-1) do
9.                    q <- m[i][k] + m[k+1][j] + p[i-1]*p[k]*p[j]
10.                   if q < m[i][j] then
11.                         m[i][j] <- q
12.                         s[i][j] <- k
13. Print m[1][n] and PrintOrder(s, 1, n)
14. Stop

Algorithm: PrintOrder(s, i, j)

1. If (i == j) then
2.        print "A"i
3. else
4.        print "("
5.        PrintOrder(s, i, s[i][j])
6.        PrintOrder(s, s[i][j]+1, j)
7.        print ")"
```

````python
import time
n=int(input("Enter the number of matrices:"))
p=list(map(int,input(f"Enter the {n+1} dimensions (matrix i is p[i-1] x p[i]):").split()))

st=time.perf_counter()
m=[[0]*(n+1) for _ in range(n+1)]
s=[[0]*(n+1) for _ in range(n+1)]
c=0
for l in range(2,n+1):
    for i in range(1,n-l+2):
        j=i+l-1
        m[i][j]=10**9
        for k in range(i,j):
            c+=1
            q=m[i][k]+m[k+1][j]+p[i-1]*p[k]*p[j]
            if q<m[i][j]: m[i][j]=q; s[i][j]=k
t=time.perf_counter()-st

def order(i,j):
    return f"A{i}" if i==j else "("+order(i,s[i][j])+order(s[i][j]+1,j)+")"

print("\nCost table:")
for i in range(1,n+1):
    print("".join("\t-" if j<i else f"\t{m[i][j]}" for j in range(1,n+1)))
print(f"Minimum number of multiplications:{m[1][n]}\nOptimal parenthesization:{order(1,n)}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(n^3)\nSpace complexity: O(n^2)")
````

## 13. 0/1 Knapsack (DP)

**Aim:** 13. Write a program in python to solve the 0/1 knapsack problem using dynamic programming and find its execution time using the time function.

```
Algorithm: Knapsack01(v, wt, n, W)

1. Start
2. For w <- 0 to W do
3.        V[0][w] <- 0
4. For i <- 1 to n do
5.        For w <- 0 to W do
6.              if wt[i] <= w and v[i] + V[i-1][w-wt[i]] > V[i-1][w] then
7.                    V[i][w] <- v[i] + V[i-1][w-wt[i]]
8.                    keep[i][w] <- 1
9.              else
10.                   V[i][w] <- V[i-1][w]
11.                   keep[i][w] <- 0
12. print "Maximum value" V[n][W]
13. w <- W
14. For i <- n down to 1 do
15.       if keep[i][w] == 1 then
16.             print "Item i selected"
17.             w <- w - wt[i]
18. Stop
```

````python
import time
n=int(input("Enter the number of items:"))
v=[0]+list(map(int,input(f"Enter the values of {n} items:").split()))
wt=[0]+list(map(int,input(f"Enter the weights of {n} items:").split()))
W=int(input("Enter the capacity of the knapsack:"))

st=time.perf_counter()
V=[[0]*(W+1) for _ in range(n+1)]
keep=[[0]*(W+1) for _ in range(n+1)]
c=0
for i in range(1,n+1):
    for w in range(W+1):
        c+=1
        if wt[i]<=w and v[i]+V[i-1][w-wt[i]]>V[i-1][w]:
            V[i][w]=v[i]+V[i-1][w-wt[i]]; keep[i][w]=1
        else: V[i][w]=V[i-1][w]
t=time.perf_counter()-st

print("\nV[i,w] table:\ni\\w"+"".join(f"\t{w}" for w in range(W+1)))
for i in range(n+1): print(str(i)+"".join(f"\t{x}" for x in V[i]))
sel,w="",W
for i in range(n,0,-1):
    if keep[i][w]: sel+=f"\tItem {i}(v={v[i]},w={wt[i]})"; w-=wt[i]
print(f"Maximum value:{V[n][W]}\nItems selected:{sel}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(nW)\nSpace complexity: O(nW)")
````

## 14. Longest Common Subsequence (DP)

**Aim:** 14. Write a program in python to find the longest common subsequence (LCS) of two sequences using dynamic programming and find its execution time using the time function.

```
Algorithm: LCS-Length(X, Y, m, n)

1. Start
2. For i <- 1 to m do
3.        c[i][0] <- 0
4. For j <- 0 to n do
5.        c[0][j] <- 0
6. For i <- 1 to m do
7.        For j <- 1 to n do
8.              if X[i] == Y[j] then
9.                    c[i][j] <- c[i-1][j-1] + 1
10.                   b[i][j] <- 'D'
11.             else if c[i-1][j] >= c[i][j-1] then
12.                   c[i][j] <- c[i-1][j]
13.                   b[i][j] <- 'U'
14.             else
15.                   c[i][j] <- c[i][j-1]
16.                   b[i][j] <- 'L'
17. print "Length of LCS" c[m][n]
18. PrintLCS(b, X, m, n)
19. Stop

Algorithm: PrintLCS(b, X, i, j)

1. If (i == 0 or j == 0) then
2.        Return
3. If b[i][j] == 'D' then
4.        PrintLCS(b, X, i-1, j-1)
5.        print X[i]
6. else if b[i][j] == 'U' then
7.        PrintLCS(b, X, i-1, j)
8. else
9.        PrintLCS(b, X, i, j-1)
```

````python
import time
x=input("Enter the first sequence X:").strip()
y=input("Enter the second sequence Y:").strip()
m,n=len(x),len(y)

st=time.perf_counter()
c=[[0]*(n+1) for _ in range(m+1)]
b=[[""]*(n+1) for _ in range(m+1)]
cnt=0
for i in range(1,m+1):
    for j in range(1,n+1):
        cnt+=1
        if x[i-1]==y[j-1]: c[i][j]=c[i-1][j-1]+1; b[i][j]="D"
        elif c[i-1][j]>=c[i][j-1]: c[i][j]=c[i-1][j]; b[i][j]="U"
        else: c[i][j]=c[i][j-1]; b[i][j]="L"
t=time.perf_counter()-st

def lcs(i,j):
    if i==0 or j==0: return ""
    if b[i][j]=="D": return lcs(i-1,j-1)+x[i-1]
    return lcs(i-1,j) if b[i][j]=="U" else lcs(i,j-1)

print("\nc[i,j] table (D=diagonal, U=up, L=left):\n\t\t"+"".join(ch+"\t" for ch in y))
for i in range(m+1):
    print((x[i-1] if i else "")+"\t"+"".join(f"{c[i][j]}{b[i][j]}\t" for j in range(n+1)))
print(f"Length of LCS:{c[m][n]}\nLCS:{lcs(m,n)}")
print(f"Number of comparisons:{cnt}\nTime taken:{t:f} seconds")
print("Time complexity: O(mn)\nSpace complexity: O(mn)")
````

## 15. Fractional Knapsack (Greedy)

**Aim:** 15. Write a program in python to solve the fractional knapsack problem using the greedy method and find its execution time using the time function.

```
Algorithm: FractionalKnapsack(v, w, n, W)

1. Start
2. For i <- 1 to n do
3.        r[i] <- v[i]/w[i]
4.        x[i] <- 0
5. Sort the items in decreasing order of r[i] (merge sort)
6. remaining <- W, total <- 0
7. For each item i in sorted order, while remaining > 0 do
8.        if w[i] <= remaining then
9.              x[i] <- 1
10.             remaining <- remaining - w[i]
11.       else
12.             x[i] <- remaining/w[i]
13.             remaining <- 0
14.       total <- total + x[i]*v[i]
15. print "Maximum value" total
16. Stop
```

````python
import time
c=0

def mergeSort(a):
    global c
    if len(a)<2: return a
    m=(len(a)+1)//2
    L,R,out=mergeSort(a[:m]),mergeSort(a[m:]),[]
    while L and R:
        c+=1
        out.append(L.pop(0) if r[L[0]]>=r[R[0]] else R.pop(0))
    return out+L+R

n=int(input("Enter the number of items:"))
v=[0]+list(map(float,input(f"Enter the values of {n} items:").split()))
w=[0]+list(map(float,input(f"Enter the weights of {n} items:").split()))
W=float(input("Enter the capacity of the knapsack:"))

st=time.perf_counter()
r=[0]+[v[i]/w[i] for i in range(1,n+1)]
x=[0]*(n+1)
o=mergeSort(list(range(1,n+1)))
rem,tot=W,0
for i in o:
    if rem<=0: break
    c+=1
    if w[i]<=rem: x[i]=1; rem-=w[i]
    else: x[i]=rem/w[i]; rem=0
    tot+=x[i]*v[i]
t=time.perf_counter()-st

print("\nItems in decreasing order of value/weight:\nItem\tValue\tWeight\tv/w\tFraction")
for i in o: print(f"{i}\t{v[i]:g}\t{w[i]:g}\t{r[i]:g}\t{x[i]:g}")
print(f"Maximum value:{tot:g}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(n log n)\nSpace complexity: O(n)")
````

## 17. Huffman Coding (Greedy)

**Aim:** 17. Write a program in python to construct an optimal prefix code using Huffman coding (greedy method) and find its execution time using the time function.

```
Algorithm: Huffman(C, f, n)

1. Start
2. Q <- min-priority queue of all n characters, keyed on frequency f
3. For i <- 1 to (n-1) do
4.        allocate a new node z
5.        x <- ExtractMin(Q)
6.        y <- ExtractMin(Q)
7.        left[z] <- x
8.        right[z] <- y
9.        f[z] <- f[x] + f[y]
10.       Insert(Q, z)
11. root <- ExtractMin(Q)
12. PrintCodes(root, "")
13. Stop

Algorithm: PrintCodes(z, code)

1. If z is a leaf then
2.        print character of z and code
3.        Return
4. PrintCodes(left[z], code + "0")

```

````python
import time
n=int(input("Enter the number of characters:"))
ch=input(f"Enter the {n} characters:").split()
f=list(map(int,input(f"Enter the frequencies of {n} characters:").split()))+[0]*n
L,R,Q,c=[-1]*(2*n),[-1]*(2*n),[],0

def heapify(i):
    global c
    s,l,r=i,2*i+1,2*i+2
    if l<len(Q):
        c+=1
        if f[Q[l]]<f[Q[s]]: s=l
    if r<len(Q):
        c+=1
        if f[Q[r]]<f[Q[s]]: s=r
    if s!=i: Q[i],Q[s]=Q[s],Q[i]; heapify(s)

def extractMin():
    m=Q[0]
    Q[0]=Q[-1]; Q.pop()
    heapify(0)
    return m

def insert(z):
    global c
    Q.append(z); i=len(Q)-1
    while i>0:
        p=(i-1)//2; c+=1
        if f[Q[i]]>=f[Q[p]]: break
        Q[i],Q[p]=Q[p],Q[i]; i=p

def printCodes(z,code):
    if L[z]==-1: print(f"{ch[z]}\t{f[z]}\t{code or '0'}"); return
    printCodes(L[z],code+"0")
    printCodes(R[z],code+"1")

st=time.perf_counter()
for i in range(n): insert(i)
for z in range(n,2*n-1):
    x,y=extractMin(),extractMin()
    L[z],R[z],f[z]=x,y,f[x]+f[y]
    insert(z)
root=extractMin()
t=time.perf_counter()-st

print("\nHuffman codes:\nChar\tFreq\tCode")
printCodes(root,"")
print(f"Total frequency at root:{f[root]}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(n log n)\nSpace complexity: O(n)")
````

## 22. Job Assignment (Branch & Bound)

**Aim:** 22. Write a program in python to solve the job assignment problem using branch and bound and find its execution time using the time function.

```
Algorithm: JOB-ASSIGNMENT(C, n)   (Least Cost Branch and Bound)
(lb(X) = cost of the jobs assigned so far in X
         + for every worker not yet assigned, his cheapest job among the jobs still free)

1. Start
2. Create the root node: no worker assigned, cost <- 0
3. lb(root) <- lower bound of root
4. Insert root into the list of live nodes
5. While the list of live nodes is not empty do
6.       E <- the live node with minimum lb
7.       Remove E from the list of live nodes
8.       if all n workers are assigned in E then
9.             print the assignment in E and its cost as the optimal solution
10.            Stop
11.      i <- the next worker to be assigned in E
12.      For each job j not yet assigned in E do
13.            Create a child X of E with worker i assigned job j
14.            cost(X) <- cost(E) + C[i][j]
15.            lb(X) <- lower bound of X
16.            Insert X into the list of live nodes
17. Stop
```

````python
import time
INF=99999
n=int(input("Enter the number of workers/jobs:"))
print(f"Enter the {n}x{n} cost matrix (row = worker, column = job):")
c=[[0]*(n+1)]+[[0]+list(map(int,input().split())) for _ in range(n)]
par,lev,job,used,cost,lb,live=[],[],[],[],[],[],[]
cnt=0

def bound(k):
    global cnt
    b=cost[k]
    for i in range(lev[k]+1,n+1):
        mn=INF
        for j in range(1,n+1):
            cnt+=1
            if not used[k]>>j&1 and c[i][j]<mn: mn=c[i][j]
        b+=mn
    return b

def newNode(p,j):
    l=lev[p]+1 if p!=-1 else 0
    par.append(p); lev.append(l); job.append(j)
    used.append(used[p]|1<<j if p!=-1 else 0)
    cost.append(cost[p]+c[l][j] if p!=-1 else 0)
    lb.append(bound(len(lb))); live.append(1)

def path(k):
    if par[k]==-1: return "root"
    return (path(par[k])+", " if par[par[k]]!=-1 else "")+f"W{lev[k]}->J{job[k]}"

st=time.perf_counter()
ordr,fst,ans=[],[],0
newNode(-1,0)
while True:
    E=-1
    for k in range(len(live)):
        if live[k]:
            cnt+=1
            if E==-1 or lb[k]<lb[E]: E=k
    if E==-1: break
    live[E]=0
    if lev[E]==n: ans=E; break
    ordr.append(E); fst.append(len(live))
    for j in range(1,n+1):
        if not used[E]>>j&1: newNode(E,j)
t=time.perf_counter()-st

print("\nNodes expanded (least lb first):")
for E,f0 in zip(ordr,fst):
    print(f"E = {{{path(E)}}}  lb = {lb[E]}")
    for k in range(f0,f0+n-lev[E]): print(f"      W{lev[k]}->J{job[k]}   cost = {cost[k]:<4} lb = {lb[k]}")
asg,k=[0]*(n+1),ans
while par[k]!=-1: asg[lev[k]]=job[k]; k=par[k]
print("\nOptimal assignment:")
for i in range(1,n+1): print(f"Worker {i} -> Job {asg[i]}\tcost = {c[i][asg[i]]}")
print(f"Minimum total cost:{cost[ans]}\nNodes generated:{len(par)}\nNodes expanded:{len(ordr)}")
print(f"Number of comparisons:{cnt}\nTime taken:{t:f} seconds")
print("Time complexity: O(n!) in the worst case, far less when the bound prunes the tree")
print("Space complexity: O(n!) in the worst case for the live nodes")
````

## 23. Dijkstra's Shortest Path (Greedy)

**Aim:** 23. Write a program in python to find the shortest path from a source vertex to all other vertices using Dijkstra's algorithm and find its execution time using the time function.

```
Algorithm: DA(G, w, s)

1. INITIALIZE-SINGLE-SOURCE(G, s)
2. S <- Ø
3. Q <- V[G]
4. while Q != Ø
5.       do u <- EXTRACT-MIN(Q)
```

````python
import time
INF=99999
n=int(input("Enter the number of vertices:"))
e=int(input("Enter the number of edges:"))
print("Enter the edges (source destination weight):")
edges=[tuple(map(int,input().split())) for _ in range(e)]
s=int(input("Enter the source vertex:"))

st=time.perf_counter()
d,p,c,neg=[INF]*(n+1),[0]*(n+1),0,False
d[s]=0
for i in range(n-1):
    for u,v,w in edges:
        c+=1
        if d[u]!=INF and d[u]+w<d[v]: d[v]=d[u]+w; p[v]=u
for u,v,w in edges:
    c+=1
    if d[u]!=INF and d[u]+w<d[v]: neg=True; break
t=time.perf_counter()-st

def path(v):
    return path(p[v])+f" -> {v}" if p[v] else str(v)

if neg: print("\nGraph contains a negative weight cycle")
else:
    print("\nVertex\tDistance\tPath")
    for v in range(1,n+1):
        print(f"{v}\tunreachable" if d[v]==INF else f"{v}\t{d[v]}\t\t{path(v)}")
print(f"Number of comparisons:{c}\nTime taken:{t:f} seconds")
print("Time complexity: O(VE)\nSpace complexity: O(V+E)")
````
